```python
# 1. Extract Width and Height using Regex
# The pattern looks for two sets of digits separated by 'x'
resolution = df['ScreenResolution'].str.extract(r'(\d+)x(\d+)')
df['Resolution_Width'] = resolution[0].astype(int)
df['Resolution_Height'] = resolution[1].astype(int)

# 2. Calculate Total Pixels
df['Pixels'] = df['Resolution_Width'] * df['Resolution_Height']

# 3. Create Binary Features (IPS, Touchscreen, Retina)
df['IPS'] = df['ScreenResolution'].str.contains('IPS').astype(int)
df['Touchscreen'] = df['ScreenResolution'].str.contains('Touchscreen').astype(int)
df['Retina'] = df['ScreenResolution'].str.contains('Retina').astype(int)

# 4. Drop original column
df = df.drop(columns=['ScreenResolution'])

# 5. Drop dimension features
df = df.drop(columns=['Resolution_Width', 'Resolution_Height'])

# 6. Display confirmation
print("Screen resolution features extracted successfully.")
display(df[['Pixels', 'IPS', 'Touchscreen', 'Retina']].head())
display(df.head())
```

```python
def refine_categorize_cpu(cpu_str):
    cpu_str = str(cpu_str).lower()
    if 'core i7' in cpu_str: return 'Core i7'
    if 'core i5' in cpu_str: return 'Core i5'
    if 'core i3' in cpu_str: return 'Core i3'
    if 'core m' in cpu_str: return 'Core M'
    if 'celeron' in cpu_str: return 'Celeron'
    if 'pentium' in cpu_str: return 'Pentium'
    if 'atom' in cpu_str: return 'Atom'
    if 'xeon' in cpu_str: return 'Xeon'
    if 'ryzen' in cpu_str: return 'Ryzen'
    if any(x in cpu_str for x in ['a-series', 'a4', 'a6', 'a8', 'a10', 'a12', 'a9']): return 'A-Series'
    if 'e-series' in cpu_str: return 'E-Series'
    if 'fx' in cpu_str: return 'FX'
    words = cpu_str.split()
    return ' '.join(words[:2]).title() if len(words) >= 2 else 'Other'

df['CPU_Family'] = df['CPU_Type'].apply(refine_categorize_cpu)
print('Refined CPU Family counts:')
print(df['CPU_Family'].value_counts())
display(df[['CPU_Type', 'CPU_Family']].head())
```

```python
import re
import numpy as np

def extract_gen_refined(cpu_str):
    # Search for any sequence of exactly 4 digits
    match = re.search(r'(\d)\d{3}', str(cpu_str))
    if match:
        return int(match.group(1))
    return np.nan

# Re-apply extraction with the refined regex
df['CPU_Generation'] = df['CPU_Type'].apply(extract_gen_refined)

# Verification
print('Refined CPU Generation extraction completed.')
print(f'Missing generation values: {df["CPU_Generation"].isnull().sum()}')
display(df[['CPU_Type', 'CPU_Generation']].head(10))
```

```python
import re

def extract_cpu_class(cpu_str):
    # Search for a 4-digit number followed immediately by letters
    match = re.search(r'\d{4}([a-zA-Z]+)', str(cpu_str))
    if match:
        return match.group(1).upper()
    return 'None'

# Apply the function to create 'CPU_Class'
df['CPU_Class'] = df['CPU_Type'].apply(extract_cpu_class)

# Display the distribution of CPU classes
print('Distribution of CPU Class suffixes:')
print(df['CPU_Class'].value_counts())

# Display first 10 rows for verification
display(df[['CPU_Type', 'CPU_Class']].head(10))
```

```python
import re
import pandas as pd

CAP_RE = re.compile(r'(\d+(?:\.\d+)?)\s*(TB|GB)', re.I)

def to_gb(value: float, unit: str, tb_base: int = 1000) -> float:
    unit = unit.upper()
    return value * tb_base if unit == "TB" else value

def detect_type(text: str) -> str | None:
    t = text.upper()

    # common SSD indicators
    if "SSD" in t:
        return "SSD"
    if "HDD" in t:
        return "HDD"
    # flash-like storage
    if "Flash Storage" in t:
        return "Flash Storage"
    if "HYBRID" in t:
        return "Hybrid"
    return None

def parse_storage(memory_str, tb_base: int = 1000) -> pd.Series:
    s = "" if memory_str is None else str(memory_str).strip()
    parts = [p.strip() for p in s.split("+") if p.strip()]

    storage = {"SSD": 0.0, "HDD": 0.0, "Flash Storage": 0.0, "Hybrid": 0.0}

    for part in parts:
        m = CAP_RE.search(part)
        if not m:
            continue

        value = float(m.group(1))
        unit = m.group(2)
        gb = to_gb(value, unit, tb_base=tb_base)

        stype = detect_type(part)
        if stype:
            storage[stype] += gb

    return pd.Series(storage)
```

```python
# 1. Apply the parsing function to the Memory column
storage_features = df['Memory'].apply(parse_storage)

# 2. Concatenate the new features to the main dataframe
df = pd.concat([df, storage_features], axis=1)

# 3. Drop the original Memory column
df = df.drop(columns=['Memory'])

print('\nSample of the final storage features:')
display(df[['SSD', 'HDD', 'Flash Storage', 'Hybrid']].tail(10))
```

```python
def extract_gpu_family(gpu_str):
    gpu_str = str(gpu_str).upper()
    if 'GTX' in gpu_str: return 'GTX'
    if 'MX' in gpu_str: return 'MX'
    if 'IRIS' in gpu_str: return 'Iris'
    if 'UHD' in gpu_str: return 'UHD'
    if 'HD' in gpu_str: return 'HD'
    if 'RADEON' in gpu_str: return 'Radeon'
    if 'QUADRO' in gpu_str: return 'Quadro'
    return 'Other'

# 1. Extract GPU Family
df['GPU_Family'] = df['GPU_Type'].apply(extract_gpu_family)

# 3. Display results
print("GPU Families extracted and One-Hot Encoded:")
display(df[['GPU_Type', 'GPU_Family']].tail(20))

# 3. Drop intermediate column (optional, but clean for ML)
#df = df.drop(columns=['GPU_Type'])
```

```python
def categorize_opsys(os_str):
    os_str = str(os_str).lower()
    if 'windows' in os_str:
        return 'Windows'
    if 'mac' in os_str or 'apple' in os_str:
        return 'Mac'
    if 'linux' in os_str:
        return 'Linux'
    if 'chrome' in os_str:
        return 'Chrome'
    if 'android' in os_str:
        return 'Android'
    if 'no os' in os_str:
        return 'No_OS'
    return 'Other'

# 1. Create the OpSys_Group column
df['OpSys_Group'] = df['OpSys'].apply(categorize_opsys)

# 2. Display the transformation and value counts
print("Operating System grouping completed:")
print(df['OpSys_Group'].value_counts())

display(df[['OpSys', 'OpSys_Group']].head(20))
```
